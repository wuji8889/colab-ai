# 免费显卡画图

用 Colab 的免费显卡（T4，约 16G 显存）画图。

**先做一步：** 点上面菜单「运行时」→「更改运行时类型」→ 选 GPU（T4），保存。

然后点「运行时」→「全部运行」，等几分钟，出来一个链接，点开就能画图。

```python
import torch
print("显卡能用吗:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("显卡型号:", torch.cuda.get_device_name(0))
```

```python
from diffusers import StableDiffusionPipeline
import torch, gradio as gr

pipe = StableDiffusionPipeline.from_pretrained(
    "Lykon/DreamShaper",
    torch_dtype=torch.float16,
).to("cuda")
pipe.safety_checker = None

def draw(prompt):
    image = pipe(prompt, num_inference_steps=25).images[0]
    return image

gr.Interface(
    fn=draw,
    inputs=gr.Textbox(label="画面描述（写英文，比如 a cat in space）"),
    outputs=gr.Image(label="画出来的图"),
    title="免费显卡画图",
).launch(share=True)
```
